Сведения о наиболее популярных женских и мужских именах среди новорожденных с 2015 по 2025


Задачи:


1.   Топ 100 популярных имен за период с 2015 по 2025 год (график)
2.   Топ 10 популярных мужских имен за каждый месяц (график или графики)
3.   Топ 10 популярных женских имен за каждый месяц (график или графики)
4.   Топ 10 популярных мужских имен за весь период (график или графики)
5.   Топ 10 популярных женских имен за весь период (график или графики)

```python
import csv
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_m = pd.read_csv('Number_of_names_man.csv', sep=';', header=1)
df_w = pd.read_csv('Number_of_names_woman.csv', sep=';', header=1)
```

Изучим данные

```python
df_m
```

```python
df_w
```

Удалим столбец `Unnamed:6` из обоих датафреймов

```python
df_m.drop(columns=['Unnamed: 6'], inplace=True)
df_w.drop(columns=['Unnamed: 6'], inplace=True)
```

```python
df_m
```

```python
df_w
```

Приведём названия месяцев к единому виду

```python
df_m['Месяц'] = df_m['Месяц'].str.capitalize()
df_w['Месяц'] = df_w['Месяц'].str.capitalize()
```

Перейдём к заданиям

## Задание 1
Топ 100 популярных имен за период с 2015 по 2025 год (график)

```python
df_m_all_years = df_m.groupby(by='Имя')['Количество человек'].sum().sort_values(ascending=False)
df_w_all_years = df_w.groupby(by='Имя')['Количество человек'].sum().sort_values(ascending=False)
```

```python
df_w_all_years
```

```python
df_w_all_years
```

```python
fig, axes = plt.subplots(1, 2, figsize=(20, 50))
axes[0].barh(y=df_m_all_years.head(100).index, width=df_m_all_years.head(100), color='blue')
axes[0].invert_xaxis()
axes[1].barh(y=df_w_all_years.head(100).index, width=df_w_all_years.head(100), color='red')
```

## Исправление задания 1
Сделаем также общий график как для мужских, так и для женских имён

```python
df_m_mark = pd.DataFrame(df_m_all_years)
df_m_mark['sex'] = 'm'
df_w_mark = pd.DataFrame(df_w_all_years)
df_w_mark['sex'] = 'w'
```

Объединим

```python
df_all_years = pd.concat((df_m_mark, df_w_mark))
df_all_years
```

```python
df_all_years = df_all_years.sort_values(by='Количество человек', ascending=False)
df_all_years['color'] = 'blue'
df_all_years.loc[df_all_years['sex'] == 'm', 'color'] = 'red'

df_all_years
```

```python
fig, ax = plt.subplots(1, 1, figsize=(20, 50))
ax.barh(y=df_all_years.head(100).index, width=df_all_years['Количество человек'].head(100), color=df_all_years.head(100)['color'])
```

## Задание 2
Топ 10 популярных мужских имен за каждый месяц (график или графики)

```python
for year in df_m['Год'].unique():
    for month in df_m[df_m['Год'] == year]['Месяц'].unique():
        data_f = df_m[(df_m['Год'] == year) & (df_m['Месяц'] == month)].groupby(by='Имя')['Количество человек'].sum().sort_values(ascending=True)
        plt.barh(y=data_f.tail(10).index, width=data_f.tail(10))
        plt.title(f'10 самых популярных мужских имён, {year}, {month}')
        plt.show()
```

Также сделаю здесь подсчёт с группировкой просто по месяцам, без годов

```python
for month in df_m['Месяц'].unique():
    data_f = df_m[df_m['Месяц'] == month].groupby(by='Имя')['Количество человек'].sum().sort_values(ascending=True)
    plt.barh(y=data_f.tail(10).index, width=data_f.tail(10))
    plt.title(f'10 самых популярных мужских имён, {month}')
    plt.show()
```

## Задание 3
Топ 10 популярных женских имен за каждый месяц (график или графики)

```python
for year in df_w['Год'].unique():
    for month in df_w[df_w['Год'] == year]['Месяц'].unique():
        data_f = df_w[(df_w['Год'] == year) & (df_w['Месяц'] == month)].groupby(by='Имя')['Количество человек'].sum().sort_values(ascending=True)
        plt.barh(y=data_f.tail(10).index, width=data_f.tail(10))
        plt.title(f'10 самых популярных женских имён, {year}, {month}')
        plt.show()
```

Также сделаю здесь подсчёт с группировкой просто по месяцам, без годов

```python
for month in df_w['Месяц'].unique():
    data_f = df_w[df_w['Месяц'] == month].groupby(by='Имя')['Количество человек'].sum().sort_values(ascending=True)
    plt.barh(y=data_f.tail(10).index, width=data_f.tail(10))
    plt.title(f'10 самых популярных женских имён, {month}')
    plt.show()
```

## Задание 4
Топ 10 популярных мужских имен за весь период (график или графики)

```python
df_m_all_years = df_m.groupby(by='Имя')['Количество человек'].sum().sort_values()
plt.barh(y=df_m_all_years.tail(10).index, width=df_m_all_years.tail(10))
plt.title(f'10 самых популярных мужских имён за всё время')
plt.show()
```

## Задание 5
Топ 10 популярных женских имен за весь период (график или графики)

```python
df_w_all_years = df_w.groupby(by='Имя')['Количество человек'].sum().sort_values()
plt.barh(y=df_w_all_years.tail(10).index, width=df_m_all_years.tail(10))
plt.title(f'10 самых популярных женских имён за всё время')
plt.show()
```
