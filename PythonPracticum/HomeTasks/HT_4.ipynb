````python
class Rectangle:
    def __init__(self, p1=Point(0, 0), p2=Point(1, 1)):
        if p1.x == p2.x or p1.y == p2.y:
            print('Невозможно создать прямоугольник с заданными координатами, инициализируем на точках (0, 0) и (1, 1)')
            self.p1 = Point(0, 0)
            self.p2 = Point(1, 1)
        else:
            self.p1 = p1
            self.p2 = p2
        
    def __str__(self):
        return f'Прямоугольник ({self.p1.x}, {self.p1.y}), ({self.p2.x}, {self.p2.y})'
    
    def perimeter(self):
        return f'Периметр: {abs(self.p1.x - self.p2.x) * 2 + abs(self.p1.y - self.p2.y) * 2}'
    
    def area(self):
        return f'Площадь: {abs((self.p1.x - self.p2.x) * (self.p1.y - self.p2.y))}'
    
    def contains(self, p3=Point()):
        if min(self.p1.x, self.p2.x) <= p3.x and p3.x <= max(self.p1.x, self.p2.x) and min(self.p1.y, self.p2.y) <= p3.y and p3.y <= max(self.p1.y, self.p2.y):
            return True
        return False
````

# Задание 6 
Доработать предыдущую задачу, чтобы можно было складывать двое часов друг с другом. Для перегрузки оператора + использовать метод __add__(self, other).

````python
class Watch:
    def __init__(self, hour=0, minute=0, second=0):
        if 0 <= hour and hour <= 23:
            self.hour = hour
        else:
            raise Exception('Неправильно указано количество часов!')
        if 0 <= minute and minute <= 59:
            self.minute = minute
        else:
            raise Exception('Неправильно указано количество минут!')
        if 0 <= second and second <= 59:
            self.second = second
        else:
            raise Exception('Неправильно указано количество секунд!')
    
    def inc_min(self):
        self.minute += 1
        if self.minute > 59:
            self.inc_hour()
            self.minute = 0

    
    def inc_hour(self):
        self.hour = (self.hour + 1) % 24
    
    def inc_sec(self):
        self.second += 1
        if self.second > 59:
            self.inc_min()
            self.second = 0
    
    def __str__(self):
        return f'Время: {self.hour:02}:{self.minute:02}:{self.second:02}'
    
    def __add__(self, other):
        if self.second + other.second > 59:
            self.inc_min()
        if self.minute + other.minute > 59:
            self.inc_hour()
        return Watch((self.hour + other.hour) % 24, (self.minute + other.minute) % 60, (self.second + other.second) % 60)
````

# Задание 7 
Создать классы для травоядного животного и травы. Животное должно уметь поедать траву, если испытывает голод, в противном случае отказываться от лакомства. Трава должна обладать питательностью, в зависимости от которой животное будет насыщаться.

````python
class grass():
    def __init__(self, richness=0):
        if richness >= 0:
            self.richness = richness
        else:
            raise Exception('Питательность не может быть отрицательной!')
    
    def consumption(self, consumption):
        if self.richness > 0:
            self.richness -= min(self.richness, consumption)
            print(f'Питательность травы теперь составляет {self.richness}')
        else:
            print('Эту траву уже не съесть')
    
    def __str__(self):
        return f'Моя питательность: {self.richness}'


class herbivore():
    def __init__(self, hunger=0):
        if hunger >= 0:
            self.hunger = hunger
        else:
            raise Exception('Голод не может быть отрицательным!')
    
    def consume(self, food=grass()):
        if self.hunger > 0:
            if food.richness > 0:
                consumed = min(self.hunger, food.richness)
                self.hunger -= consumed
                food.consumption(consumed)
                print(f'Я утолил голод на {consumed}, теперь голод {self.hunger}')
            else:
                print('Такой травой я не наемся')
        else:
            print('Не хочу есть')
    
    def __str__(self):
        return f'Мой голод: {self.hunger}'
````

- найдите потенциальные источники ошибок (укажите номера строк в строке документации);
- используя конструкцию try добавьте в код обработку соответствующих исключений.

````python
class NoMoneyToWithdrawError(Exception):
    def __init__(self, message):
        super().__init__(message)

class PaymentError(Exception):
    def __init__(self, message):
        super().__init__(message)

def print_accounts(accounts):
    """Печать аккаунтов.
    
    Потенциальные ошибки:
    - строка 12: TypeError при передаче не-dict
    - строка 13: KeyError при отсутствии ключей в accounts
    """
    try:
        print("Список клиентов ({}): ".format(len(accounts)))
        for i, (name, value) in enumerate(accounts.items(), start=1):
            print("{}. {} {}".format(i, name, value))
    except TypeError:
        print("Ошибка: передан некорректный тип данных")
    except KeyError:
        print("Ошибка: отсутствуют ожидаемые ключи в словаре")

def transfer_money(accounts, account_from, account_to, value):
    """Выполнить перевод 'value' денег со счета 'account_from' на 'account_to'.
    
    При переводе денежных средств необходимо учитывать:
    - хватает ли денег на счету, с которого осуществляется перевод;
    - перевод состоит из уменьшения баланса первого счета и увеличения
      баланса второго; если хотя бы на одном этапе происходит ошибка,
      аккаунты должны быть приведены в первоначальное состояние
      (механизм транзакции)
      см. https://ru.wikipedia.org/wiki/Транзакция_(информатика).
    
    Исключения (raise):
    - NoMoneyToWithdrawError: на счету 'account_from'
                               не хватает денег для перевода;
    - PaymentError: ошибка при переводе.
    """
    # Потенциальные ошибки:
    # - строка 50: KeyError при отсутствии аккаунта в словаре
    # - строка 52: TypeError при некорректном типе value
    # - строка 55: ValueError при отрицательном значении value
    # - строка 57: NoMoneyToWithdrawError при недостатке средств
    # - строка 76: PaymentError при общей ошибке перевода
    
    try:
        if account_from not in accounts:
            raise KeyError(f'Счет {account_from} не найден')
        if account_to not in accounts:
            raise KeyError(f'Счет {account_to} не найден')
            
        if not isinstance(value, (int, float)):
            raise TypeError('Сумма должна быть числом')
        if value <= 0:
            raise ValueError('Сумма должна быть положительной')
            
        original_from = accounts[account_from]
        original_to = accounts[account_to]
        
        if accounts[account_from] < value:
            raise NoMoneyToWithdrawError(
                f'Недостаточно средств на счете {account_from}')
            
        accounts[account_from] -= value
        accounts[account_to] += value
        
    except (KeyError, TypeError, ValueError) as e:
        raise PaymentError(f'Ошибка при переводе: {str(e)}')
    except NoMoneyToWithdrawError as e:
        raise NoMoneyToWithdrawError(str(e))
    except Exception as e:
        accounts[account_from] = original_from
        accounts[account_to] = original_to
        raise PaymentError(f'Непредвиденная ошибка: {str(e)}')
````

````python
def convert_marks():
    print('Программа перевода оценок.\nДля выхода просто нажмите Enter без ввода значения\nВводите оценки по-одной на строку')
    marks_dict = {}
    i = 0
    while True:
        user_input = input('Введите оценку: ').strip()
        
        if user_input == '':
            print(f'\nПрограмма завершена.\n')
            return marks_dict
            
        try:
            number = float(user_input)
            
            if number > 4.5 and number <= 5:
                i += 1
                print(f'{i}. {number}: A')
                marks_dict[f'{i}. {number}'] = 'A'
            elif number <= 4.5 and number > 4:
                i += 1
                print(f'{i}. {number}: B')
                marks_dict[f'{i}. {number}'] = 'B'
            elif number <= 4 and number > 3.5:
                i += 1
                print(f'{i}. {number}: C')
                marks_dict[f'{i}. {number}'] = 'C'
            elif number <= 3.5 and number > 3:
                i += 1
                print(f'{i}. {number}: D')
                marks_dict[f'{i}. {number}'] = 'D'
            elif number <= 3 and number > 2.5:
                i += 1
                print(f'{i}. {number}: E')
                marks_dict[f'{i}. {number}'] = 'E'
            elif number <= 2.5 and number >= 0:
                i += 1
                print(f'{i}. {number}: F')
                marks_dict[f'{i}. {number}'] = 'F'
            else:
                print(f'{number} не входит в допустимый диапазон числовых оценок')
            
        except ValueError:
            ui_u = user_input.upper()
            if ui_u == 'A':
                i += 1
                print(f'{i}. {user_input}: 5')
                marks_dict[f'{i}. {user_input}'] = '5'
            elif ui_u == 'B':
                i += 1
                print(f'{i}. {user_input}: 4+')
                marks_dict[f'{i}. {user_input}'] = '4+'
            elif ui_u == 'C':
                i += 1
                print(f'{i}. {user_input}: 4')
                marks_dict[f'{i}. {user_input}'] = '4'
            elif ui_u == 'D':
                i += 1
                print(f'{i}. {user_input}: 3+')
                marks_dict[f'{i}. {user_input}'] = '3+'
            elif ui_u == 'E':
                i += 1
                print(f'{i}. {user_input}: 3')
                marks_dict[f'{i}. {user_input}'] = '3'
            elif ui_u == 'F':
                i += 1
                print(f'{i}. {user_input}: 2')
                marks_dict[f'{i}. {user_input}'] = '2'
            else:
                print(f'{user_input} не является приемлимой оценкой по системе ECTS')
            continue
````
