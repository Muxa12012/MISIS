Простой пример: стек из книг на столе. Единственной книгой, чья обложка видна, является самая верхняя. Чтобы получить доступ к, например, третьей снизу книге, нам нужно убрать все книги, лежащие сверху, одну за другой.

````python
class Stack:
    def __init__(self, *args):
        self._stack = [*args]
    
    def push(self, elem):
        self._stack.append(elem)
    
    def pop(self):
        if len(self._stack) == 0:
            print('Стэк пустой!')
        else:
            elem = self._stack[-1]
            self._stack = self._stack[:-1]
            return elem
    
    def __str__(self):
        a = ''
        for i in range(len(self._stack)):
            a += str(self._stack[i])
            if i + 1 < len(self._stack):
                a += '; '
        return a
````

````python
class TaskManager:
    def __init__(self):
        self.task_level = {}
    
    def new_task(self, task, level):
        if level in self.task_level:
            self.task_level[level].push(task.lower())
        else:
            self.task_level[level] = Stack(task.lower())
        myKeys = list(self.task_level.keys())
        myKeys.sort()
        sd = {i: self.task_level[i] for i in myKeys}
        self.task_level = sd
    
    def __str__(self):
        a = f'Результат:\n'
        for x, y in self.task_level.items():
            a += f'{x}: {y}\n'
        return a
````

# Задача 2. Кэширование запросов
## Контекст

Вы разрабатываете программу для кэширования запросов к внешнему API. Часто повторяющиеся запросы занимают много времени, поэтому вы решаете создать класс LRU Cache (Least Recently Used Cache), который будет хранить ограниченное количество запросов и автоматически удалять самые старые при достижении лимита. Это позволит значительно ускорить повторяющиеся запросы, так как данные будут браться из кэша, а не отправляться повторно.

## Задача

Создайте класс LRU Cache, который хранит ограниченное количество объектов и, при превышении лимита, удаляет самые давние (самые старые) использованные элементы.

Реализуйте методы добавления и извлечения элементов с использованием декораторов property и setter.
```python
@property
def cache(self): # этот метод должен возвращать самый старый элемент
    ...
@cache.setter
def cache(self, new_elem): # этот метод должен добавлять новый элемент
    ...
```
## Советы

Не забывайте обновлять порядок использованных элементов. В итоге должны удаляться давно использованные элементы, а не давно добавленные, так как давно добавленный элемент может быть популярен, и его удаление не поможет ускорить новые запросы.
## Пример:
```python
# Создаём экземпляр класса LRU Cache с capacity = 3
cache = LRUCache(3)
```

````python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity: int):
        if capacity <= 0:
            raise ValueError("Capacity must be positive integer")
        self._capacity = capacity
        self._cache = OrderedDict()  # Используем OrderedDict для сохранения порядка вставки

    @property
    def cache(self):
        # этот метод должен возвращать самый старый элемент
        if not self._cache:
            return None
        return next(iter(self._cache.items()))

    @cache.setter
    def cache(self, new_elem):
        if not isinstance(new_elem, tuple) or len(new_elem) != 2:
            raise TypeError("Element must be a tuple of (key, value)")
        
        key, value = new_elem
        
        # Если ключ уже существует, обновляем значение и перемещаем в конец (делаем самым новым)
        if key in self._cache:
            self._cache[key] = value
            self._cache.move_to_end(key)
        else:
            # Если кэш переполнен, удаляем самый старый элемент (первый)
            if len(self._cache) >= self._capacity:
                self._cache.popitem(last=False)  # last=False удаляет первый элемент
            # Добавляем новый элемент (он становится самым новым — в конце)
            self._cache[key] = value

    def get(self, key):
        if key not in self._cache:
            return None
        
        # Перемещаем элемент в конец (делаем самым новым по использованию)
        self._cache.move_to_end(key)
        return self._cache[key]

    def __len__(self):
        # Возвращает текущее количество элементов в кэше.
        return len(self._cache)

    def __contains__(self, key):
        # Проверяет наличие ключа в кэше.
        return key in self._cache

    def clear(self):
        # Очищает кэш.
        self._cache.clear()

    def keys(self):
        # Возвращает список ключей в порядке от самого старого к самому новому.
        return list(self._cache.keys())

    def values(self):
        # Возвращает список значений в порядке от самого старого к самому новому.
        return list(self._cache.values())

    def items(self):
        # Возвращает список пар (ключ, значение) в порядке от самого старого к самому новому.
        return list(self._cache.items())

    def __repr__(self):
        return f"LRUCache({self._capacity}, {dict(self._cache)})"
````

````python
# Создаём кэш с лимитом в 3 элемента
cache = LRUCache(3)

# Добавляем элементы
cache.cache = ("key1", "value1")
cache.cache = ("key2", "value2")
cache.cache = ("key3", "value3")

print(cache)

# Получаем самый старый элемент
print(cache.cache)

# Обращаемся к среднему элементу - он становится самым новым
print(cache.get("key2"))
print(cache)

# Добавляем новый элемент - удаляется самый старый (key1)
cache.cache = ("key4", "value4")
print(cache)

# Проверяем наличие ключа
print("key2" in cache)
print("key1" in cache) 
````

# Задача 3. Кэширование для ускорения вычислений
## Контекст
Вы разрабатываете программу для оптимизации вычислений чисел Фибоначчи. Числа Фибоначчи вычисляются рекурсивной функцией, каждое число равно сумме двух предыдущих чисел. Однако вы заметили, что при больших значениях чисел Фибоначчи вычисления занимают значительное время, так как многие значения вычисляются повторно. Вам поручено создать декоратор, который кэширует результаты вызова функции и позволяет избежать повторных вычислений для одних и тех же аргументов.

## Задача
Создайте декоратор, который кэширует (сохраняет для дальнейшего использования) результаты вызова функции и, при повторном вызове с теми же аргументами, возвращает сохранённый результат.

Примените его к рекурсивной функции вычисления чисел Фибоначчи.

В итоге декоратор должен проверять аргументы, с которыми вызывается функция, и, если такие аргументы уже использовались, должен вернуть сохранённый результат вместо запуска расчёта.

## Советы
- Для хранения результатов удобно использовать словарь, так как поиск элементов внутри словаря будет иметь сложность, равную в среднем O(1).
- При этом не стоит хранить все вычисления в одном словаре, созданном снаружи функций (в глобальной области видимости). Лучше создавать
 отдельные словари для каждой декорируемой функции.

````python
def cache_decorator(func):
    # Создаём словарь для хранения результатов для функции
    cache = {}
    
    def wrapper(*args, **kwargs):
        # Формируем ключ на основе аргументов вызова
        # Для простоты работаем только с позиционными аргументами (числа Фибоначчи принимают только один аргумент)
        key = args
        
        # Проверяем, есть ли результат в кэше
        if key in cache:
            return cache[key]
        
        # Если результата нет в кэше — вычисляем и сохраняем
        result = func(*args, **kwargs)
        cache[key] = result
        return result
    
    return wrapper


@cache_decorator
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
````

````python
# Пример использования
if __name__ == "__main__":
    # Вычисляем несколько чисел Фибоначчи
    print(f"F(10) = {fibonacci(10)}")
    print(f"F(20) = {fibonacci(20)}")
    print(f"F(30) = {fibonacci(30)}")
    
    # Повторный вызов с тем же аргументом — результат берётся из кэша
    print(f"F(10) (повторно) = {fibonacci(10)}")
````

# Задача 4. Крестики нолики

Напишите программу, которая реализует игру Крестики-нолики. 

Ваши классы в этой задаче могут выглядеть так:
```python
class Cell:
    #  Клетка, у которой есть значения
    #   - занята она или нет
    #   - номер клетки

class Board:
    #  Класс поля, который создаёт у себя экземпляры клетки

class Player:
    #  У игрока может быть
    #   - имя
    #   - на какую клетку ходит

````python
class Cell:
    def __init__(self, x, marked='@'):
        try:
            x = int(x)
            if x <= 0:
                raise ValueError('Клетка может располагаться только на числах от 1 до 9')
            else:
                self.x = x
        except ValueError:
            print('Ошибка, это не число!')
        
        if marked in ['X', 'O', '@']:
            self.marked = marked
        else:
            raise ValueError('Клетка может быть либо пустой (@), либо X, либо O')
        
    def is_marked(self):
        if self.marked == 'X' or self.marked == 'O':
            return True
        else:
            return False
    
    def mark(self, symb):
        self.marked = symb
    
    def __str__(self):
        return self.marked
````

````python
class Board:
    def __init__(self):
        self.board_map = [Cell(i + 1) for i in range(9)]
    
    def is_full(self):
        return all(cell.is_marked() for cell in self.board_map)
    
    def __str__(self):
        ret = ''
        ret = '  1 | 2 | 3\n'
        ret += f'1 {self.board_map[0]} | {self.board_map[1]} | {self.board_map[2]}\n'
        ret += f'2 {self.board_map[3]} | {self.board_map[4]} | {self.board_map[5]}\n'
        ret += f'3 {self.board_map[6]} | {self.board_map[7]} | {self.board_map[8]}\n'
        return ret
    
    def win_cond(self, symb):
        # Все возможные комбинации клеток для поля 3х3
        wins = [[0, 1, 2],
                [6, 7, 8],
                [0, 3, 6],
                [1, 4, 7],
                [2, 5, 8],
                [0, 4, 8],
                [2, 4, 6]]
        for cond in wins:
            if all(self.board_map[i].marked == symb for i in cond):
                return True
        return False
    
    def set_mark(self, x, y, symb):
        if self.board_map[(y - 1) * 3 + x - 1].is_marked():
            print('Клетка уже занята!')
            return False
        else:
            self.board_map[(y - 1) * 3 + x - 1].mark(symb)
            return True
````

````python
class Player:
    def __init__(self, name, symb):
        self.name = name
        if symb not in ('X', 'O'):
            raise ValueError('Символ игрока либо X, либо O')
        else:
            self.symb = symb
    
    def make_move(self):
        while True:
            try:
                print(f'{self.name}, введите номер клетки через запятую:')
                x, y = [int(i) for i in input().split(', ')]
                if 1 <= x <= 3 and 1 <= y <= 3:
                    print(f'x = {x}, y = {y}')
                    return x, y
                else:
                    print('Клетка за пределами игрового поля!')
            except ValueError:
                print('Введите корректный номер клетки!')
    
    def __str__(self):
        return f'Игрок: {self.name}, символ: {self.symb}'
````

````python
def play():
    print('Начинаем игру в крестики-нолики!')
    import random
    name1 = input('Введите имя первого игрока:')
    name2 = input('Введите имя второго игрока:')
    choice = 3
    while choice not in [1, 2]:
        choice = int(input('Случайный выбор первого игрока (1) или по порядку (2):'))
    if choice == 1:
        print('Случайный выбор')
        first = random.randint(1, 2)
        if first == 1:
            print(f'Первым начинает игрок {name1}!')
            player1 = Player(name1, 'X')
            player2 = Player(name2, 'O')
        else:
            print(f'Первым начинает игрок {name2}!')
            player1 = Player(name2, 'X')
            player2 = Player(name1, 'O')
    else:
        print('По очереди')
        print(f'Первым начинает игрок {name1}!')
        player1 = Player(name1, 'X')
        player2 = Player(name2, 'O')
    
    board = Board()

    print(f'Первый игрок: {player1}')
    print(f'Второй игрок: {player2}')

    cur_player = player1

    while True:
        print(board)

        cell_x, cell_y = cur_player.make_move()

        if board.set_mark(cell_x, cell_y, cur_player.symb):
            if board.win_cond(cur_player.symb):
                print(board)
                print(f'{cur_player.name} выиграл!')
                break

            if board.is_full():
                print(board)
                print('Ничья!')
                break

            cur_player = player2 if cur_player == player1 else player1
````
