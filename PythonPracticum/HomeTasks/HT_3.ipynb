# Задание 1
Дан список из чисел.
Определите их НОК (наименьшее общее кратное) и НОД (наибольший общий делитель).

````python
def gcd(a, b):

    a = abs(a)
    b = abs(b)

    while b != 0:
        a, b = b, a % b
    
    return a
````

````python
def gcd_list(numbers):
    if not numbers:
        return None
    
    result = numbers[0]
    
    for num in numbers[1:]:
        result = gcd(result, num)
        
        if result == 1:
            break
    
    return result
````

````python
def lcm(a, b):
    if a == 0 or b == 0:
        return 0
    
    return abs(a * b) // gcd(a, b)
````

````python
def lcm_list(numbers):
    if not numbers:
        return None
    
    result = 1
    
    for num in numbers:
        result = lcm(result, num)
        
        if result == 0:
            break
    
    return result
````

````python
numbers = [int(x) for x in input('Введите список чисел через пробел: ').split(' ')]

print(numbers)

gcd_result = gcd_list(numbers)
print(f"НОД чисел {numbers} = {gcd_result}")  # НОД = 6

lcm_result = lcm_list(numbers)
print(f"НОК чисел {numbers} = {lcm_result}")  # НОК = 360
````

````python
import string
# не будем кодировать букву ё, она неудачно расположена в юникоде
def encode(word, num):
    res = ''
    for i in word:
        if (i <= 'Я' and i >= 'А') or (i <= 'я' and i >= 'а'):
            if i.islower():
                res += chr((ord(i) + num - ord('а')) % 32 + ord('а'))
            else:
                res += chr((ord(i) + num - ord('А')) % 32 + ord('А'))
        else:
            res += i
    return res

def decode(word, num):
    res = ''
    for i in word:
        if (i <= 'Я' and i >= 'А') or (i <= 'я' and i >= 'а'):
            if i.islower():
                res += chr((ord(i) - num - ord('а')) % 32 + ord('а'))
            else:
                res += chr((ord(i) - num - ord('А')) % 32 + ord('А'))
        else:
            res += i
    return res

sentence = 'Слово было, слова не стало!'
n = 3
sentence = encode(sentence, 3)
print(sentence, n)
sentence = decode(sentence, n)
print(sentence, 3)
````

отрицательных значений (отсортирован по убыванию);
неотрицательных значений (отсортирован по возрастанию).


````python
def func(*args):
    neg = []
    zero_pos = []
    for i in args:
        if i < 0:
            neg.append(i)
        else:
            zero_pos.append(i)
    neg = sorted(neg, reverse=True)
    zero_pos = sorted(zero_pos)
    return(neg, zero_pos)

func(1, 2, 3, 4, 5, -7, 0, -5, 3, 4, -8)
````

# Задание 7
Строка называется палиндромом, если она пишется одинаково в обоих направлениях. Например, палиндромами в английском языке являются слова «anna», «civic», «level», «hannah». Напишите программу, запрашивающую у пользователя строку и при помощи цикла определяющую, является ли она палиндромом.


````python
def palindrom(word):
    for i in range(len(word) // 2):
        if word[i] != word[-i - 1]:
            return False
    return True
````

````python
print(palindrom('w'))
print(palindrom('wa'))
print(palindrom('waaw'))
print(palindrom('waraw'))
print(palindrom('waspaw'))
````

Вы отвечаете одним из трёх чисел:
- 1 — равно,
- 2 — больше,
- 3 — меньше.

Напишите программу, которая с помощью цепочки таких вопросов и ответов угадывает число.

Дополнительно: сделайте так, чтобы можно было гарантированно угадать число за семь попыток.

````python
import numpy as np
def guess(num):
    print(f'Загадано число {num}')
    nums = np.linspace(1, 100, 100)
    flag = 0
    n = 0
    while flag != 1:
        n += 1
        check = nums[int(len(nums) / 2)]
        flag = int(input(f'Твое число равно, меньше или больше, чем число {check}?'))
        print(f'Попытка №{n}: {check}')
        if flag == 2:
            nums = [i for i in nums if i > check]
            print('Больше')
        elif flag == 3:
            nums = [i for i in nums if i < check]
            print('Меньше')
    print('Равно')
    return (check, n)
````

Напишите программу, которая позволит пользователю преобразовывать числа из одной системы счисления в другую произвольным образом.

Ваша программа должна поддерживать все системы счисления в диапазоне от 2 до 16 как для входных, так и для выходных данных. Если пользователь выберет систему с основанием, выходящим за границы допустимого, на экран должна быть выведена ошибка.

Разделите код программы на несколько функций, включая функцию, конвертирующую число из произвольной системы счисления в десятичную, и обратную функцию, переводящую значение из десятичной системы в произвольную. В основной программе необходимо запросить у пользователя исходную систему счисления, целевую систему, а также число для преобразования.


````python
hex_dict = {i: chr(i + 55) if i > 9 else str(i) for i in range(16)}
hex_dict_rev = {j: i for i, j in hex_dict.items()}

def translate(num, orig, fin):
    def to_dec(num, orig):
        num = str(num)
        res = 0
        for i in range(len(num)):
            res += hex_dict_rev[num[i]] * orig ** (len(num) - i - 1)
        return res

    def from_dec(num, fin):
        res = ''
        while num > 0:
            a = num % fin
            num = num // fin
            res = hex_dict[a] + res
        return res

    return from_dec(to_dec(num, orig), fin)
````

# Задание 10

Магическими называются даты, в которых произведение дня и месяца составляет последние две цифры года. Например, 10 июня 1960 года – магическая дата, поскольку 10 ´ 6 = 60.

Напишите функцию, определяющую, является ли введенная дата магической. Используйте написанную функцию в главной программе для отображения всех магических дат в XX ве­ке.

````python
import datetime as dt
````

````python
def is_magic(date_check):
    if type(date_check) != dt.date:
        date_check = dt.date.fromisoformat(date_check)
    day = date_check.day
    month = date_check.month
    year = date_check.year
    if day * month == year % 100:
        return True
    return False
````

````python
is_magic('1960-06-10')
````

````python
datec = dt.date.fromisoformat('1901-01-01')
cnt = 0
while datec.year < 2001:
    if is_magic(datec):
        print(f'{datec} - магическая дата')
        cnt += 1
    datec += dt.timedelta(1)
print(f'Всего {cnt} магических дат')
````
